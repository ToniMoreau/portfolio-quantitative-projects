#### quick clean

```python
data = data.drop(columns=["code_parent", "echelle_geo", "nb_ventes_local", "moy_prix_m2_local", "med_prix_m2_local"])
```

```python
data.columns
```

#### Real data cleaning/organisation

```python
paris_data = data[data["code_geo"] == "75"]
paris_data["annee_mois"] = pd.to_datetime(paris_data["annee_mois"])
paris_data["mois"] = paris_data["annee_mois"].dt.month 
paris_data = paris_data.sort_values("annee_mois")

paris_all_data = paris_data[["code_geo", "nb_ventes_apt_maison","moy_prix_m2_apt_maison", "med_prix_m2_apt_maison", "libelle_geo", "annee_mois", "mois"]]
paris_all_data["taux de variation"] = paris_all_data["moy_prix_m2_apt_maison"].pct_change()
paris_all_data.head()
```

```python
paris_maison_data = paris_data[["code_geo", "nb_ventes_maison","moy_prix_m2_maison", "med_prix_m2_maison", "libelle_geo", "annee_mois", "mois"]]
paris_maison_data["taux de variation"] = paris_maison_data["moy_prix_m2_maison"].pct_change()
paris_maison_data.head()
```

```python
paris_appartement_data = paris_data[["code_geo", "nb_ventes_appartement","moy_prix_m2_appartement", "med_prix_m2_appartement", "libelle_geo", "annee_mois", "mois"]]
paris_appartement_data["taux de variation"] = paris_appartement_data["moy_prix_m2_appartement"].pct_change()
paris_appartement_data = paris_appartement_data.dropna(subset=['taux de variation'])
paris_appartement_data.head()
```

> sensitivity is too high due to few transactions for Paris house market. Thus we will only focus on appartment

#### final results

```python
r_month = paris_appartement_data["taux de variation"].mean()
sigma_month = paris_appartement_data["taux de variation"].std()

print(r_month, sigma_month)
```
